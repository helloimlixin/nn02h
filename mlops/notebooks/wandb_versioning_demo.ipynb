# Versioning with Weights & Biases

```python
import wandb
```

## Create an artifact

```python
with open("test_artifact.txt", "w+") as f:
    f.write("This is an example artifact.")
```

```python
%ls
```

## Create a run and Upload the artifact

```python
run = wandb.init(project="wandb_artifact_demo",
                 group="experiment1")
```

```python
artifact = wandb.Artifact(name="test_artifact.txt", # good practice to name the artifact the same as the file
                          type="data", description="A test artifact.",
                          metadata={"key1": "val1"})
```

```python
artifact.add_file("test_artifact.txt")  # does not need to be a single file
```

```python
run.log_artifact(artifact)
```

```python
run.finish()  # finish the run to upload the artifact
```

## Create A New Version of the Artifact

```python
with open("test_artifact.txt", "w+") as f:
    f.write("This is an updated example artifact.")

run = wandb.init(project="wandb_artifact_demo1",
                    group="experiment2")

artifact = wandb.Artifact(name="test_artifact.txt",
                          type="data", description="A test artifact.",
                          metadata={"key2": "val2"})

artifact.add_file("test_artifact.txt")
run.log_artifact(artifact)
```

```python
run.finish()
```

```python
# re-upload the original artifact
with open("test_artifact.txt", "w+") as f:
    f.write("This is an updated example artifact.")

run = wandb.init(project="wandb_artifact_demo1",
                    group="experiment2")

artifact = wandb.Artifact(name="test_artifact.txt",
                          type="data", description="A test artifact.",
                          metadata={"key2": "val2"})

artifact.add_file("test_artifact.txt")
run.log_artifact(artifact)

run.finish()
```

## Context Managers

```python
with wandb.init(project="multiple_runs") as run:
    artifact = wandb.Artifact(name="test_artifact.txt",
                              type="data", description="A test artifact.",
                              metadata={"key1": "val1"})

    artifact.add_file("test_artifact.txt")
```

With the context manager, you don't need to call `run.finish()` as it is called automatically when the context manager exits. This is useful for ensuring that the artifact is uploaded even if an exception is raised. Also, we can run multiple runs in the same script without needing to worry about finishing the run using the context manager.
